## Warm-up: Lists, Loops and Functions (5 minutes)

Before introducing new data structures, let's recap three building blocks from earlier weeks:

- **List:** An ordered collection of values written inside square brackets, such as `[9, 11, 14, 8]`.
- **`for` loop:** A control structure that repeats an indented block of code once for each item in a sequence.
- **Function:** A named, reusable block of code defined with `def`. A function can accept inputs called **parameters** inside its parentheses and send a result back to the caller using the `return` keyword.

Run the example cell below to see all three working together.

```python
temperatures = [9, 11, 14, 8]

def count_mild_days(temp_list):
    count = 0
    for temp in temp_list:
        if temp >= 10:
            count = count + 1
    return count

mild_days = count_mild_days(temperatures)
print("Mild days:", mild_days)
```

## Lab 1: Dictionaries (25 minutes)

### Part 1: Formatting text with f-strings

Next week's weather app displays formatted forecast summaries in the terminal. Concatenating strings and numbers with `+` and `str()` gets cluttered quickly.

An **f-string** (short for *formatted string literal*) is a string with the letter `f` placed immediately before the opening quotation mark. Inside an f-string, you can place variables or Python expressions inside curly braces `{}`. Python evaluates each `{}` and inserts its value into the text automatically, converting numbers to text for you.

Run the cell below to see an f-string in action.

```python
day_name = "Friday"
high_temp = 11
low_temp = 4

summary = f"{day_name} will reach a high of {high_temp}°C and a low of {low_temp}°C."
print(summary)
```

### Part 2: Creating and looking up values in a dictionary

Suppose we store one day's weather forecast in a list: `["Friday", 11, 4, 60]`. To use that list, you have to memorise that index `1` is the high temperature, index `2` is the low temperature, and index `3` is the percentage chance of rain.

A **dictionary** solves this by storing data as labelled **key-value pairs** inside curly braces `{}`:
- **Key:** A unique label (most often a string, such as `"high"`) that identifies a piece of data.
- **Value:** The data associated with that key (such as `11`). Each key is separated from its value by a colon `:`, and pairs are separated by commas.
- **Lookup:** To retrieve a value from a dictionary, write the dictionary's name followed by the key inside **square brackets**: `forecast["high"]`.
- **Membership check (`in`):** Writing `key in dictionary` returns `True` if `key` is one of the keys in the dictionary, and `False` otherwise.

Run the example cell below.

```python
forecast = {"day": "Friday", "high": 11, "low": 4, "rain_chance": 60}

print("Day:", forecast["day"])
print("High temperature:", forecast["high"])
print("Has rain_chance key?", "rain_chance" in forecast)
print("Has wind_speed key?", "wind_speed" in forecast)
```

### Part 3: Changing values, adding keys, and `KeyError`

Dictionaries are **mutable**, meaning you can change their contents after creating them:
- **Changing an existing value:** Assign a new value to an existing key in square brackets, such as `forecast["high"] = 13`.
- **Adding a new key-value pair:** Assign a value to a brand-new key in square brackets, such as `forecast["wind_mph"] = 15`.
- **`KeyError`:** An error Python raises when you try to look up a key in square brackets that does not exist in the dictionary. This often happens when a key name is misspelled.
- **Square brackets only:** Dictionaries always use square brackets `forecast["high"]` to look up keys, never dot notation (`forecast.high`).

Run the example cell below.

```python
forecast = {"day": "Friday", "high": 11, "low": 4, "rain_chance": 60}

# Update an existing key's value
forecast["rain_chance"] = 75

# Add a brand-new key-value pair
forecast["wind_mph"] = 18

print(forecast)
```

### Part 4: Looping through a dictionary with `.items()`

To loop through both the keys and the values of a dictionary at the same time, we use `.items()`.

`.items()` is a **method**—a function that belongs to a specific object (here, our dictionary) and is called after a dot `.`. When you loop over `forecast.items()`, each step of the `for` loop gives you two things: the **key** and its **value**.

Run the example cell below.

```python
forecast = {"day": "Friday", "high": 11, "low": 4, "rain_chance": 60}

for key, value in forecast.items():
    print(f"Key: {key} -> Value: {value}")
```

### Part 1: What are objects and methods?

You have already used **objects** and **methods** throughout this course:
- When you wrote `wet_hours.append("12:00")`, `wet_hours` was a list object and `.append()` was a method on that list.
- When you wrote `"Hello world".split()`, the text was a string object and `.split()` was a method on that string.
- When you wrote `forecast.items()`, `forecast` was a dictionary object and `.items()` was a method on that dictionary.

Here are the key terms for working with objects:
- **Object:** A container that bundles data and the functions that work on that data into a single unit.
- **Attribute (also called state):** A variable stored inside an object, accessed with a dot (for example, `friday.high`).
- **Method:** A function that belongs to an object and can read or change that object's attributes, called with a dot (for example, `friday.describe()`).

#### Why build our own objects instead of using dictionaries?
With a dictionary, the data (`forecast`) and the functions that use it (`describe_forecast(forecast)`) are separate:
1. Nothing guarantees that a dictionary has all the right keys spelled properly (`"high"` vs `"hihg"`).
2. Anyone could accidentally pass the wrong dictionary into a function.
3. Your code editor cannot autocomplete `forecast["..."]` keys, whereas typing `friday.` lets the editor suggest attributes and methods automatically.

### Part 2: Defining a class (`__init__` and `self`)

To make our own objects, we first write a **class**:
- **Class:** A blueprint or template that defines what attributes and methods a category of objects will have. By Python convention (PEP 8), class names use `CapWords` (capitalising the start of each word without underscores), such as `DayForecast`.
- **Instance:** An individual object built from a class. Creating a new object from a class is called **instantiating** the class.
- **Constructor (`__init__`):** A special method named `__init__` (short for *initialise*, written with two underscores before and two underscores after). Python runs `__init__` automatically whenever you create a new instance, sets up its starting attributes, and automatically returns the new object.
- **`self`:** The first parameter in `__init__` and every other method inside a class. Python automatically sets `self` to refer to the specific object being created or used. Writing `self.high = high` attaches an attribute named `high` onto that object.

Run the example cell below to see a `DayForecast` class and two instances created from it.

```python
class DayForecast:
    def __init__(self, day, high, low, rain_chance):
        self.day = day
        self.high = high
        self.low = low
        self.rain_chance = rain_chance

# Create (instantiate) two separate objects from the DayForecast class
friday = DayForecast("Friday", 11, 4, 60)
saturday = DayForecast("Saturday", 16, 8, 10)

# Look up attributes on each object using dot notation
print(friday.day, "high:", friday.high)
print(saturday.day, "high:", saturday.high)
```

Notice the difference between dictionaries and custom objects:
- On a **dictionary**, we look up keys with square brackets: `forecast["high"]`.
- On an **object** created from a class, we access attributes with a dot: `friday.high`.

### Part 3: Writing methods on `DayForecast`

Now let's give our `DayForecast` objects actions—called **methods**—by defining functions indented inside the `class DayForecast:` block.

- Every method must list `self` as its first parameter so it can access the object's attributes (`self.day`, `self.high`, `self.low`, `self.rain_chance`).
- When you call a method on an object, such as `friday.temp_spread()`, Python passes `friday` into the `self` parameter automatically, so you do not write anything inside the parentheses `()`.

Run the example cell below.

```python
class DayForecast:
    def __init__(self, day, high, low, rain_chance):
        self.day = day
        self.high = high
        self.low = low
        self.rain_chance = rain_chance

    def temp_spread(self):
        return self.high - self.low

friday = DayForecast("Friday", 11, 4, 60)
saturday = DayForecast("Saturday", 16, 8, 10)

print("Friday spread:", friday.temp_spread())
print("Saturday spread:", saturday.temp_spread())
```

### Extension C: Making `print(forecast)` work with `__str__`

Try running `print(friday)` right now—Python prints something like `<__main__.DayForecast object at 0x...>`, because Python does not know how you want a `DayForecast` formatted as text.

A **dunder method** (short for *double underscore* method, also called a *special method*) is a method with two underscores before and after its name. Python calls dunder methods automatically in specific situations:
- `__init__(self, ...)` is called automatically when you create a new object.
- `__str__(self)` is called automatically whenever you pass an object to `print(obj)` or `str(obj)`. It must **return** a string.

Complete the `?` placeholder in `__str__(self)` below so that it returns the string produced by `self.describe()`. Expect `print(friday)` to display `Friday: 11°C / 4°C (60% rain)` and `Correct: True`.

```python
class DayForecast:
    def __init__(self, day, high, low, rain_chance):
        self.day = day
        self.high = high
        self.low = low
        self.rain_chance = rain_chance

    def describe(self):
        return f"{self.day}: {self.high}°C / {self.low}°C ({self.rain_chance}% rain)"

    def needs_coat(self):
        if self.high < 12 or self.rain_chance >= 50:
            return True
        else:
            return False

    def __str__(self):
        return self.?()

friday = DayForecast("Friday", 11, 4, 60)
saturday = DayForecast("Saturday", 16, 8, 10)

print(friday)
print(saturday)
print("Correct:", str(friday) == "Friday: 11°C / 4°C (60% rain)")
```
