````python
class DayForecast:
    def __init__(self, day, high, low, rain_chance):
        self.day = day
        self.high = high
        self.low = low
        self.rain_chance = rain_chance

    def describe(self):
        return f"{self.day}: {self.high}°C / {self.low}°C ({self.rain_chance}% rain)"

    def needs_coat(self):
        if self.high < 12 or self.rain_chance >= 50:
            return True
        else:
            return False

    def is_warm(self):
        if self.high >= 15:
            return True
        else:
            return False
````
